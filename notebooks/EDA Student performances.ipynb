```python
numerical_features = [feature for feature in df.columns if df[feature].dtype != "O"]
categorical_features = [feature for feature in df.columns if df[feature].dtype == "O"]

print(f"We have {len(numerical_features)} numerical features")
print(f"We have {len(categorical_features)} categorical features")
```

```python
df.head()
```

```python
df["total_score"] = df["math score"] + df["reading score"] + df["writing score"]
df["avarage_score"] = df["total_score"]/3
```

```python
df.head()
```

```python
math_full_score = df[df["math score"] ==100]["avarage_score"].count()
reading_full_score = df[df["reading score"] == 100]["avarage_score"].count()
writing_full_score = df[df["writing score"]==100]["avarage_score"].count()
print("math_full_score is ",math_full_score)
print("reading_full_score is ",reading_full_score)
print("writing_full_score is ",writing_full_score)
```

```python
less_than_20_reading = df[df["reading score"] <= 20]["avarage_score"].count()
less_than_20_math = df[df["math score"]<= 20]["avarage_score"].count()
less_than_20_writing = df[df["writing score"]<=20]["avarage_score"].count() 
print("less_than_20_reading is",less_than_20_reading)
print("less_than_20_math is",less_than_20_math)
print("less_than_20_writing is" , less_than_20_writing)
```
